```python
import pandas as pd
import pickle
import matplotlib.pyplot as plt
from bertopic import BERTopic

# Load results
results_df = pd.read_csv('results_summary.csv', index_col=0)
results_df
```

```python
# Load groups and models
with open('data/groups_preprocessed.pkl', 'rb') as f:
    groups = pickle.load(f)

topic_models = {}
for name in groups.keys():
    topic_models[name] = BERTopic.load(f'models/bertopic_{name.lower()}')

print("Models loaded")
```

```python
# Explore topics for a specific group
group_name = 'All'  # Change to 'Education', 'Humanities', 'Medicine'
model = topic_models[group_name]

topic_info = model.get_topic_info()
topic_info_filtered = topic_info[topic_info['Topic'] != -1].head(20)
topic_info_filtered
```

```python
# Get detailed topic keywords
topic_id = 0  # Change to explore different topics
words = model.get_topic(topic_id)
print(f"Topic {topic_id} keywords:")
for word, score in words[:15]:
    print(f"  {word}: {score:.4f}")
```

```python
# Visualize topic distribution
fig = model.visualize_barchart(top_n_topics=10)
fig.show()
```

```python
# Get sample documents for a topic
with open('models/topics_dict.pkl', 'rb') as f:
    topics_dict = pickle.load(f)

topic_id = 0
group_name = 'All'

group_df = groups[group_name]
topics = topics_dict[group_name]

# Find posts in this topic
topic_posts = group_df[topics == topic_id]['cleaned_text'].head(5).tolist()

print(f"Sample posts from Topic {topic_id}:")
for i, post in enumerate(topic_posts, 1):
    print(f"\n{i}. {post[:200]}...")
```
